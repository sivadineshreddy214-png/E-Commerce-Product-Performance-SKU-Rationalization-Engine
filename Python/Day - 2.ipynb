### Day-2

#### Restate the problem and success metrics

### Problem:
- Analyse order delivery performance to identify factors causing delays and improve on-time deliveries.


### Success Metrics :
- Increase on-time delivery rate.
- Reduce average delivery delay.
- Identify key factors affecting delivery performance



### Enumerate Data Sets

#### File Name
Order_Delivery_Performance_Dataset (2)(6).csv	

#### File Type
CSV	

#### Description
Contains order, customer, shipping, delivery, and logistics information for analyzing delivery performance.	

#### Purpose
Primary dataset used for exploratory data analysis (EDA), data validation, visualization, and business insights.

### Map 21 approved dataset columns to business meaning

| **Column Name**            | **Business Meaning**                                                                                                          |
-------------------------- ----------------------------------------------------------------------------------------------------------------------------
**Order_ID**                : Unique identifier assigned to each customer order. Used to track and reference orders.                                        
**Customer_ID**             : Unique identifier for each customer. Used to analyze customer purchasing and delivery behavior.                              
**Order_Date**              : Date on which the customer placed the order. Indicates when the order process began.                                          
**Processing_Time_Hours**   : Number of hours taken to process the order before shipment. Measures warehouse or order-processing efficiency.                
**Shipping_Date**           : Date the order was shipped from the warehouse. Marks the beginning of the delivery process.                                   
**Expected_Delivery_Date**  : Promised delivery date communicated to the customer. Used to measure service-level performance.                               
**Actual_Delivery_Date**    : Date the customer actually received the order. Used to determine delivery performance.                                        
**Delivery_Time_Days**      : Total number of days taken from shipping to delivery. Measures delivery speed.                                                
**Delivery_Status**         : Indicates whether the order was delivered on time, delayed, or had another delivery outcome.                                 
**Delay_Days**              : Number of days the delivery was delayed beyond the expected delivery date. Used to quantify delivery delays.                  
**Cancellation**            : Indicates whether the order was cancelled (Yes/No). Helps analyze order completion rates.                                     
**Cancellation_Reason**     : Reason provided for cancelling an order (if applicable). Helps identify common causes of cancellations.                       
**Courier**                 : Shipping or logistics company responsible for delivering the order. Used to compare courier performance.                      
**Warehouse**               : Warehouse or fulfillment center from which the order was dispatched. Used to evaluate warehouse efficiency.                   
**Region**                  : Geographic region where the order was delivered. Supports regional performance analysis.                                      
**Shipping_Cost**           : Cost incurred to ship the order. Used to analyze logistics expenses and profitability.                                        
**Order_Value**             : Total monetary value of the customer's order. Used for revenue and profitability analysis.                                    
**Customer_Rating**         : Rating given by the customer after delivery. 

**Payment_Method**          : Method used by the customer to pay for the order (e.g., Credit Card, UPI, Cash on Delivery). Supports payment trend                                    analysis.

**Product_Category**        : Category of products ordered (e.g., Electronics, Clothing, Grocery). Used to analyze product-specific delivery                                         performance.   
**Priority**                : Priority level assigned to the order (e.g., High, Medium, Low). Indicates the urgency of order fulfillment.
