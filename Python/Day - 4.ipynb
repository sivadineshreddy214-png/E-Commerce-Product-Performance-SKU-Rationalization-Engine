### Detect data-type issues (numeric-as-text, malformed dates in Order_Date, Processing_Time_Hours)


```python
import pandas as pd

# Load dataset
df = pd.read_csv("Order_Delivery_Performance_Dataset (2).csv")

print("========== CLEANING PLAN ==========\n")

# Missing Values
print("Missing Values:")
print(df.isnull().sum())

# Duplicate Rows
print("\nDuplicate Rows:", df.duplicated().sum())

# Data Types
print("\nData Types:")
print(df.dtypes)

# Unique values in categorical columns (for category harmonisation)
print("\nCategorical Columns:")
cat_cols = df.select_dtypes(include='object').columns

for col in cat_cols:
    print(f"\n{col}")
    print(df[col].unique())
```

### Detect invalid values / out-of-range in Shipping_Cost, Order_Value


```python
# Detect invalid values in Shipping_Cost
invalid_shipping = df[
    (df["Shipping_Cost"].isnull()) |
    (df["Shipping_Cost"] < 0)
]

# Detect invalid values in Order_Value
invalid_order = df[
    (df["Order_Value"].isnull()) |
    (df["Order_Value"] < 0)
]

# Display results
print("Invalid or Out-of-Range Values in Shipping_Cost")
print(invalid_shipping[["Shipping_Cost"]])

print("\nNumber of Invalid Shipping_Cost Values:", len(invalid_shipping))

print("\n-------------------------------------------\n")

print("Invalid or Out-of-Range Values in Order_Value")
print(invalid_order[["Order_Value"]])

print("\nNumber of Invalid Order_Value Values:", len(invalid_order))
```

### Detect inconsistent categories in Delivery_Status, Region, Product_Category


```python
# Columns to check
columns = ["Delivery_Status", "Region", "Product_Category"]

for col in columns:
    print(f"\n========== {col} ==========")

    # Display unique values
    print("Unique Categories:")
    print(df[col].unique())

    # Display frequency of each category
    print("\nCategory Counts:")
    print(df[col].value_counts(dropna=False))
```
