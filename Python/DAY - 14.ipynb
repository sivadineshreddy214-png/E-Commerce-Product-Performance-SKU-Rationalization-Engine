```python
import pandas as pd
```

```python
df = pd.read_csv("Order_Delivery_Performance_Dataset (2).csv")
```

### Descriptive stats for Shipping_Cost, Order_Value


```python
Shipping_stats = df["Shipping_Cost"].describe()
```

```python
print("Descriptive Statistics - Shipping Cost")
print(Shipping_stats)
```

```python
order_stats = df["Order_Value"].describe()

print("Descriptive Statistics - Order Value")
print(order_stats)
```

### Distribution Analysis

```python
import matplotlib.pyplot as plt
```

```python
plt.figure(figsize=(8,5))

plt.hist(df["Shipping_Cost"],
         bins=20,
         edgecolor="black")

plt.title("Distribution of Shipping Cost")
plt.xlabel("Shipping Cost")
plt.ylabel("Frequency")

plt.show()
```

```python
plt.figure(figsize=(8,5))

plt.hist(df["Order_Value"],
         bins=20,
         edgecolor="black")

plt.title("Distribution of Order Value")
plt.xlabel("Order Value")
plt.ylabel("Frequency")

plt.show()
```

```python
shipping_skew = df["Shipping_Cost"].skew()

print("Skewness of Shipping Cost:", shipping_skew)
```

```python
order_skew = df["Order_Value"].skew()

print("Skewness of Order Value:", order_skew)
```

```python
shipping_kurtosis = df["Shipping_Cost"].kurt()

print("kurtosis of Shipping Cost:", shipping_kurtosis)
```

```python
order_kurtosis = df["Order_Value"].kurt()

print("Kurtosis of Order Value:", order_kurtosis)
```

```python
summary = pd.DataFrame({
    "Mean":[df["Shipping_Cost"].mean(), df["Order_Value"].mean()],
    "Median": [df["Shipping_Cost"].median(), df["Order_Value"].median()],
    "Standard Deviation": [df["Shipping_Cost"].std(), df["Order_Value"].std()],
    "Skewness": [df["Shipping_Cost"].skew(), df["Order_Value"].skew()],
    "Kurtosis": [df["Shipping_Cost"].kurt(), df["Order_Value"].kurt()]
},
index=["Shipping_Cost","Order_Value"])

print(summary)
```

### Document Sampling Caveats

1. The dataset may not represent all customer orders or all regions
2. Seasonal trends may affect order values and shipping cost.
3. Outliers can influence summary statistics such as the mean
4. Mising or duplicate records may impact the analysis if not cleaned
5. The results are based only on the available sample ansd should not be generalized without additional data.
