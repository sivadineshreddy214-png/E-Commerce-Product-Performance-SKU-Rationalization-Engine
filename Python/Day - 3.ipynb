```python
import pandas as pd
```

```python
df = pd.read_csv("Order_Delivery_Performance_Dataset (2).csv")
```

### 1. Row and column counts

```python
print("Rows:", df.shape[0])
print("Columns:", df.shape[1])
```

### 2. Data types

```python
print("\nData Types:")
print(df.dtypes)
```

### 3. Missing value counts

```python
print("\nMissing Values:")
print(df.isnull().sum())
```

### 4. Duplicate records based on Order_ID and Customer_ID

```python
duplicate_count = df.duplicated(subset=["Order_ID", "Customer_ID"]).sum()

print("\nDuplicate Records (Order_ID + Customer_ID):", duplicate_count)

# Optional: Display duplicate rows if any exist
duplicates = df[df.duplicated(subset=["Order_ID", "Customer_ID"], keep=False)]

if duplicate_count > 0:
    print("\nDuplicate Rows:")
    print(duplicates)
else:
    print("\nNo duplicate records found.")
```
