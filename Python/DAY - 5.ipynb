### Author cleaning plan (missing values, duplicates, type fixes, category harmonisation)


````python
import pandas as pd
import numpy as np

# ==============================
# Step 1: Load Dataset
# ==============================
df = pd.read_csv("Order_Delivery_Performance_Dataset (2).csv")

print("Original Shape:", df.shape)

# ==============================
# Step 2: Remove Duplicate Rows
# ==============================
duplicates = df.duplicated().sum()
print("Duplicate Rows:", duplicates)

df = df.drop_duplicates()

# ==============================
# Step 3: Remove Extra Spaces
# ==============================
for col in df.select_dtypes(include='object').columns:
    df[col] = df[col].str.strip()

# ==============================
# Step 4: Standardize Text Values
# (Convert categorical values to lowercase)
# ==============================
for col in df.select_dtypes(include='object').columns:
    df[col] = df[col].str.lower()

# ==============================
# Step 5: Handle Missing Values
# ==============================
for col in df.columns:

    if df[col].dtype == 'object':
        # Replace missing categorical values with mode
        if df[col].isnull().sum() > 0:
            df[col].fillna(df[col].mode()[0], inplace=True)

    else:
        # Replace missing numerical values with median
        if df[col].isnull().sum() > 0:
            df[col].fillna(df[col].median(), inplace=True)

# ==============================
# Step 6: Convert Data Types
# ==============================

# Convert date columns (if any)
for col in df.columns:
    if "date" in col.lower():
        df[col] = pd.to_datetime(df[col], errors='coerce')

# Convert numeric columns stored as text
for col in df.columns:
    if df[col].dtype == 'object':
        try:
            df[col] = pd.to_numeric(df[col])
        except:
            pass

# ==============================
# Step 7: Remove Impossible Values
# (Example: Negative numbers)
# ==============================
numeric_cols = df.select_dtypes(include=np.number).columns

for col in numeric_cols:
    df = df[df[col] >= 0]

# ==============================
# Step 8: Check Missing Values Again
# ==============================
print("\nRemaining Missing Values")
print(df.isnull().sum())

# ==============================
# Step 9: Final Dataset Information
# ==============================
print("\nDataset Info")
print(df.info())

print("\nCleaned Shape:", df.shape)

# ==============================
# Step 10: Save Cleaned Dataset
# ==============================
df.to_csv("Cleaned_Order_Delivery_Performance_Dataset.csv", index=False)

print("\nCleaning completed successfully!")
print("Cleaned dataset saved as 'Cleaned_Order_Delivery_Performance_Dataset.csv'")
````

# Decide Keep / Drop / Impute Rule for Each Affected Column

Based on the cleaned version of the **Order Delivery Performance Dataset**, the following decisions were made:

| Column Status | Decision | Reason |
|---------------|----------|--------|
| Columns with no missing values | **Keep** | These columns contain complete and valid data and require no further action. |
| Columns with missing values that were imputed during cleaning | **Impute** | Missing numerical values were replaced using the median, and missing categorical values were replaced using the mode to preserve the dataset. |
| Columns with duplicate records | **Keep (after removing duplicates)** | Duplicate rows were removed during the cleaning process to ensure data integrity. |
| Columns requiring data type corrections | **Keep (after type conversion)** | Date columns were converted to datetime format, and numeric columns stored as text were converted to numeric types. |
| Categorical columns with inconsistent values | **Keep (after harmonisation)** | Text values were standardized by trimming spaces, converting to lowercase, and correcting inconsistent formatting. |
| Columns with excessive missing values (>40%) | **Drop (if any)** | No columns exceeded the threshold for removal. Therefore, no columns were dropped. |
